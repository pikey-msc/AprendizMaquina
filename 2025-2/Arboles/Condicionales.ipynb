 # Resumen Árboles de Clasificación y Regresión (CART)

## Introducción

Los Árboles de Clasificación y Regresión (CART, por sus siglas en inglés) son una familia de algoritmos utilizados en aprendizaje de máquina supervisado. CART es un método basado en la partición recursiva del espacio de características para construir árboles de decisión, los cuales pueden emplearse tanto para clasificación como para regresión.

---

## 1. Definición Formal

Un árbol de decisión $ T $ es una estructura jerárquica que particiona el espacio de entrada $ \mathcal{X} \subset \mathbb{R}^p $ en regiones disjuntas mediante divisiones sucesivas. El modelo toma la forma de una función:

$$ f: \mathbb{R}^p \to \mathbb{R} $$

o bien,

$$ f: \mathbb{R}^p \to \{1,2,...,K\} $$

dependiendo de si la tarea es de regresión o clasificación.

Cada nodo interno representa una condición de decisión basada en una característica específica, y cada hoja contiene un valor de predicción (para regresión) o una etiqueta de clase (para clasificación).

---

## 2. Construcción del Árbol (Algoritmo CART)

La construcción de un árbol CART sigue un procedimiento basado en la minimización de la impureza de las particiones. El proceso se divide en los siguientes pasos:

### 2.1 Criterio de Partición

Dado un conjunto de datos $ \mathcal{D} = \{(x_i, y_i)\}_{i=1}^{n} $, se busca la mejor división de un nodo en dos subconjuntos $ \mathcal{D}_1 $ y $ \mathcal{D}_2 $ mediante una regla de decisión basada en una característica $ j $ y un umbral $ s $:

$$
R_1(j,s) = \{ x \in \mathbb{R}^p \mid x_j \leq s \}, \quad R_2(j,s) = \{ x \in \mathbb{R}^p \mid x_j > s \}.
$$

La mejor partición es aquella que minimiza una función de impureza $ I(R) $, que depende del tipo de problema:

- **Para clasificación:** se usa la impureza de Gini o la entropía.
- **Para regresión:** se usa la suma de errores cuadráticos.

### 2.2 Criterios de Selección de División

#### a) Impureza de Gini (para clasificación)

La impureza de Gini se define como:

$$ G(R) = \sum_{k=1}^{K} p_k(1 - p_k), $$

donde $ p_k $ es la proporción de elementos en la región $ R $ que pertenecen a la clase $ k $.

#### b) Entropía (para clasificación)

Otra opción para medir la impureza es la entropía:

$$ H(R) = -\sum_{k=1}^{K} p_k \log p_k. $$

#### c) Error cuadrático medio (para regresión)

Para la regresión, el criterio de partición minimiza la varianza dentro de cada región. La función de costo es:

$$ C(R) = \sum_{i \in R} (y_i - \bar{y}_R)^2, $$

donde $ \bar{y}_R $ es el valor medio de las observaciones en $ R $.

### 2.3 Crecimiento y Poda del Árbol

El árbol se expande hasta que no se puedan realizar más divisiones útiles. Para evitar el sobreajuste, se emplea la poda utilizando validación cruzada y un parámetro de penalización $ \alpha $ para controlar la complejidad del árbol.

El costo de un árbol $ T $ con tamaño $ |T| $ se define como:

$$ C_{\alpha}(T) = \sum_{m=1}^{|T|} C(R_m) + \alpha |T|. $$

Se elige el árbol $ T^* $ que minimiza esta función usando validación cruzada.

---

## 3. Ejemplos de Uso

### 3.1 Clasificación con Árboles CART

Dado un conjunto de datos de clientes con las variables:
- **Edad** (numérica)
- **Ingresos** (numérica)
- **Compra de un producto** (Sí/No, variable objetivo)

Un árbol de decisión puede aprender reglas como:
- Si "Edad $< 30$" y "Ingresos $< 50K$", entonces "No compra".
- Si "Edad $\geq 30$" y "Ingresos $\geq 50K$", entonces "Compra".

### 3.2 Regresión con Árboles CART

Para predecir el precio de una casa en función de:
- **Tamaño** (m²)
- **Número de habitaciones**
- **Ubicación** (categórica)

El árbol segmenta los datos en regiones homogéneas, por ejemplo:
- Si "Tamaño $> 200$ m²" y "Ubicación = Zona A", entonces "Precio = 500,000".
- Si "Tamaño $\leq 200$ m²" y "Ubicación = Zona B", entonces "Precio = 250,000".

---

## 4. Conclusión

Los árboles CART son modelos flexibles y fáciles de interpretar para problemas de clasificación y regresión. Aunque pueden sobreajustar los datos si no se regulan adecuadamente, su capacidad de segmentar los datos en regiones homogéneas los convierte en una herramienta poderosa para el aprendizaje de máquina.


# Árboles de Decisión Condicionales

## Introducción

Los **árboles de decisión condicionales** son una extensión de los árboles de decisión tradicionales. Se utilizan para la clasificación y regresión, incorporando pruebas estadísticas en la selección de las divisiones (splits), lo que los hace más robustos en presencia de variables correlacionadas o datos desbalanceados.

Estos árboles son implementados en **Conditional Inference Trees (C-Trees)** y son una alternativa a los árboles de decisión clásicos como CART (Classification and Regression Trees). Se desarrollan a partir del marco de los modelos de inferencia condicional y están disponibles en bibliotecas como `party` en R y `sklearn.tree` en Python.

## ¿Cómo funcionan los árboles de decisión condicionales?

1. **Selección de variables basada en pruebas de hipótesis**  
   - A diferencia de CART, que selecciona la variable de división con base en criterios como Gini o entropía, los árboles condicionales usan pruebas de hipótesis para determinar la variable más significativa en cada nodo.
   - Se usa una prueba de independencia entre la variable respuesta y las predictoras.
   - Se selecciona la variable con el valor p más pequeño (menor a un umbral $\alpha$).
  
2. **Determinación de puntos de corte**  
   - Una vez seleccionada la mejor variable, se buscan los puntos óptimos de división a través de pruebas estadísticas (como pruebas de permutación).

3. **Crecimiento del árbol**  
   - El proceso de selección y partición se repite en cada nodo hasta cumplir un criterio de detención, como:
     - Número mínimo de observaciones en un nodo.
     - Significancia estadística de la división.
     - Profundidad máxima del árbol.

4. **Poda y validación**  
   - A diferencia de CART, que usa poda post hoc, los árboles condicionales integran la poda implícita basada en pruebas de significancia.

## Ventajas de los Árboles Condicionales

- **Manejan automáticamente la multicolinealidad**, ya que seleccionan variables basadas en pruebas de independencia.
- **No sufren de sobreajuste** tan fácilmente como los árboles tradicionales.
- **Poda integrada** gracias a las pruebas de hipótesis.
- **Funcionan bien con variables categóricas y continuas.**

# Árboles Condicionales: Teoría, Algoritmo y Aplicaciones

## 1. Introducción a los Árboles Condicionales

Los **Árboles Condicionales** (Conditional Inference Trees) son una variante de los árboles de decisión tradicionales que utilizan pruebas de hipótesis basadas en teoría estadística para realizar divisiones en los nodos. Fueron introducidos por **Hothorn, Hornik y Zeileis (2006)** y forman parte de un marco más general de modelos de árboles denominado **"conditional inference framework"**.

A diferencia de los árboles de decisión tradicionales (CART), que utilizan medidas como el índice de Gini o la reducción de varianza para seleccionar la mejor partición, los árboles condicionales emplean un enfoque basado en pruebas de independencia condicional, evitando así el sesgo hacia variables con más niveles y mejorando la interpretación de los resultados.


## ¿Cómo funcionan los árboles de decisión condicionales?

1. **Selección de variables basada en pruebas de hipótesis**  
   - A diferencia de CART, que selecciona la variable de división con base en criterios como Gini o entropía, los árboles condicionales usan pruebas de hipótesis para determinar la variable más significativa en cada nodo.
   - Se usa una prueba de independencia entre la variable respuesta y las predictoras.
   - Se selecciona la variable con el valor $p$ más pequeño (menor a un umbral $\alpha$).
  
2. **Determinación de puntos de corte**  
   - Una vez seleccionada la mejor variable, se buscan los puntos óptimos de división a través de pruebas estadísticas (como pruebas de permutación).

3. **Crecimiento del árbol**  
   - El proceso de selección y partición se repite en cada nodo hasta cumplir un criterio de detención, como:
     - Número mínimo de observaciones en un nodo.
     - Significancia estadística de la división.
     - Profundidad máxima del árbol.

4. **Poda y validación**  
   - A diferencia de CART, que usa poda post hoc, los árboles condicionales integran la poda implícita basada en pruebas de significancia.

---

## 2. Fundamentos Matemáticos de los Árboles Condicionales

Sea un conjunto de datos $\mathcal{D} = \{(\mathbf{x}_i, y_i)\}_{i=1}^{n}$ donde:

- $\mathbf{x}_i = (x_{i1}, x_{i2}, ..., x_{ip})$ representa el vector de características para la observación $i$.
- $y_i$ es la variable objetivo (puede ser categórica o continua).
- $n$ es el número total de observaciones.

El proceso de construcción del árbol se basa en las siguientes ideas fundamentales:

### 2.1 Pruebas de Independencia

Para cada posible variable predictora $X_j$, se realiza una prueba de hipótesis de independencia condicional entre la variable objetivo $Y$ y $X_j$:

$$ H_0: Y \perp X_j \quad \text{(Y es independiente de $X_j$)} $$

Si la hipótesis nula se rechaza (es decir, hay una relación estadísticamente significativa), entonces $X_j$ es un candidato para dividir el nodo.

Para evaluar esta independencia, se utiliza una **prueba de asociación basada en el estadístico de permutación**:

$$ p_j = \text{p-value de la prueba de independencia condicional entre } Y \text{ y } X_j. $$

La variable con el menor valor de $p_j$ es seleccionada para realizar la división.

### 2.2 Criterio de División

Una vez que se selecciona la variable $X_j^*$ con la asociación más significativa, se encuentra el punto de corte óptimo $s^*$ que maximiza la diferencia entre los grupos resultantes. La división se realiza de la siguiente forma:

$$
\begin{cases}
X_j^* \leq s^* & \Rightarrow \text{Izquierda} \\
X_j^* > s^* & \Rightarrow \text{Derecha}
\end{cases}
$$

Para datos categóricos, la división se basa en combinaciones de categorías en dos subconjuntos óptimos.

### 2.3 Control del Sobreajuste

Para evitar el sobreajuste, se utiliza un criterio de parada basado en valores de **p-value ajustados** o en un umbral de significancia predefinido ($\alpha$). Si ninguna variable tiene un $p_j$ menor que el umbral, el nodo no se divide más.

---

## 3. Algoritmo de Construcción del Árbol Condicional

1. **Inicialización**: Se parte del conjunto de datos original $\mathcal{D}$.
2. **Selección de Variable**: Se calcula el valor de p para cada variable predictora y se selecciona la que tenga el menor p-valor.
3. **División Óptima**: Se encuentra el mejor punto de corte para dividir el nodo.
4. **Recursividad**: Se repiten los pasos 2 y 3 para cada subgrupo generado, hasta que no haya divisiones significativas.
5. **Criterio de Parada**: Se detiene el crecimiento del árbol cuando:
   - No hay ninguna variable con una relación estadísticamente significativa con $Y$.
   - Se ha alcanzado un tamaño mínimo de nodo.
   - Se ha llegado a una profundidad máxima establecida.

---

## 4. Aplicaciones de los Árboles Condicionales

Los árboles condicionales pueden utilizarse tanto para **clasificación** como para **regresión**, dependiendo de la naturaleza de la variable objetivo.

- **Clasificación**: Si $Y$ es categórica, los nodos terminales del árbol asignan una clase basándose en la moda de las observaciones en cada hoja.
- **Regresión**: Si $Y$ es continua, cada nodo terminal predice la media (o mediana) de los valores de $Y$ en ese nodo.

## Ventajas de los Árboles Condicionales

- **Manejan automáticamente la multicolinealidad**, ya que seleccionan variables basadas en pruebas de independencia.
- **No sufren de sobreajuste** tan fácilmente como los árboles tradicionales.
- **Poda integrada** gracias a las pruebas de hipótesis.
- **Funcionan bien con variables categóricas y continuas.**

## Comparación con Árboles de Decisión Tradicionales (CART)

| Característica          | Árboles Condicionales | Árboles CART |
|------------------------|----------------------|-------------|
| **Criterio de división** | Pruebas de hipótesis basadas en p-values | Gini / Entropía (clasificación) / MSE (regresión) |
| **Manejo de sesgo** | No sesgado hacia variables con más niveles | Puede sesgarse hacia variables con más niveles |
| **Criterio de parada** | Basado en significancia estadística (p-value ajustado) | Basado en pureza del nodo o profundidad máxima |
| **Interpretabilidad** | Mayor debido al enfoque basado en pruebas estadísticas | Menor, especialmente si requiere poda |
| **Computacionalmente eficiente** | Más costoso debido a las pruebas estadísticas | Más rápido en general |
| **Adecuado para** | Datos con alto riesgo de sesgo en variables categóricas | Problemas estándar de clasificación y regresión |


Finalmente, Los Árboles Condicionales ofrecen una alternativa poderosa a los árboles tradicionales al basar su criterio de división en pruebas estadísticas. Esto los hace más adecuados para problemas donde la interpretación y el control del sesgo son críticos. Aunque su implementación puede ser más compleja, brindan una herramienta valiosa en aplicaciones científicas y empresariales.

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from statsmodels.sandbox.stats.multicomp import multipletests
from scipy.stats import chi2_contingency
from sklearn import datasets
import pandas as pd

# Cargar dataset iris desde scikit-learn
iris = datasets.load_iris()

# Convertirlo en un DataFrame de Pandas
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['Species'] = iris.target

# Mapear los valores numéricos a nombres de especies
df['Species'] = df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})


X = df.drop(columns=["Species"])
y = df["Species"]

# División del conjunto de datos
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Función para calcular pruebas de independencia
def independence_tests(X, y):
    p_values = []
    for col in X.columns:
        table = pd.crosstab(X[col], y)
        _, p, _, _ = chi2_contingency(table)
        p_values.append(p)
    return np.array(p_values)

# Aplicación de la prueba y selección de la variable más significativa
p_vals = independence_tests(X_train, y_train)
selected_feature = X_train.columns[np.argmin(p_vals)]
print(f"Variable seleccionada: {selected_feature}")

````
