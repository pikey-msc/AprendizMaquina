# **Práctica: Modelos Lineales Generalizados (MLG) en Datos de Accidentes de Tránsito**

## **1. Introducción**
Los **Modelos Lineales Generalizados (MLG)** permiten modelar relaciones entre variables cuando la variable dependiente no sigue una distribución normal. En esta práctica, aplicaremos MLG para analizar un conjunto de datos reales de accidentes de tránsito en México.



## **2. Conjunto de Datos**
El INEGI ofrece la Estadística de Accidentes de Tránsito Terrestre en Zonas Urbanas y Suburbanas (ATUS), que recopila información anual sobre la siniestralidad del transporte terrestre a nivel nacional, por entidad federativa y municipio. Este conjunto de datos incluye variables como la fecha y hora del accidente, tipo de accidente, características de las víctimas, condiciones de la vía, entre otras.​

Enlace al conjunto de datos: Accidentes de Tránsito Terrestre en Zonas Urbanas y Suburbanas
