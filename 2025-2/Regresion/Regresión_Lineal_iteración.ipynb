# Fundamentos de la Regresión Lineal

## 1. Función de Pérdida de Mínimos Cuadrados

En regresión lineal, la **función de pérdida** más común es la **suma de los errores al cuadrado** (OLS, por sus siglas en inglés "Ordinary Least Squares"). El objetivo es minimizar la **diferencia entre las predicciones del modelo y los valores observados**.

La función de pérdida se expresa como:

$$ L(\beta) = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{n} (y_i - (\beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \dots + \beta_p x_{ip}))^2 $$

Donde:
- $ y_i $ son los valores observados,
- $ \hat{y}_i $ son las predicciones del modelo,
- $ \beta_0, \beta_1, ..., \beta_p $ son los coeficientes del modelo,
- $ x_{i1}, x_{i2}, ..., x_{ip} $ son los valores de las variables predictoras.

El objetivo es encontrar los coeficientes $ \beta_0, \beta_1, ..., \beta_p $ que minimicen la suma de los cuadrados de los errores (residuos).

## 2. Estimador de los Coeficientes

Los coeficientes $ \beta_0, \beta_1, ..., \beta_p $ de la regresión lineal se pueden estimar mediante **mínimos cuadrados ordinarios**. El estimador de los coeficientes es:

$$
\hat{\beta} = (X^T X)^{-1} X^T y
$$

Donde:
- $ X $ es la matriz de características (con una columna de unos para el intercepto),
- $ y $ es el vector de las observaciones,
- $ \hat{\beta} $ es el vector de coeficientes estimados.

Este es un resultado directo derivado de minimizar la función de pérdida.

## 3. Distribución de los Coeficientes

Cuando se asume que los errores $ \epsilon_i $ (es decir, la diferencia entre las observaciones y las predicciones) siguen una distribución normal con media cero y varianza constante $ \sigma^2 $, los coeficientes estimados $ \hat{\beta} $ son **insesgados** y **eficientes**. La distribución de los coeficientes estimados es:

$$
\hat{\beta} \sim N(\beta, \sigma^2 (X^T X)^{-1})
$$

Esto significa que, bajo estas condiciones, los estimadores de los coeficientes siguen una distribución normal con una media igual al valor real de los coeficientes $ \beta $ y una varianza que depende de la matriz $ (X^T X)^{-1} $.

## 4. Pruebas de Hipótesis Comunes

En regresión lineal, es común realizar pruebas de hipótesis para evaluar la significancia de los coeficientes estimados. Aquí te presento algunas de las más comunes:

### a. Prueba de Significancia Individual (Prueba t)

Se utiliza para evaluar si un coeficiente estimado $ \hat{\beta_j} $ es significativamente diferente de cero. La hipótesis nula es:

$$
H_0: \beta_j = 0
$$

Contra la hipótesis alternativa:

$$
H_a: \beta_j \neq 0
$$

La estadística de prueba es:

$$
t_j = \frac{\hat{\beta_j}}{SE(\hat{\beta_j})}
$$

Donde $ SE(\hat{\beta_j}) $ es el error estándar de $ \hat{\beta_j} $. Si el valor p asociado a esta prueba es menor que el nivel de significancia (por ejemplo, 0.05), se rechaza $ H_0 $, indicando que el predictor es significativo.

### b. Prueba F Global (Prueba de la Significancia Global del Modelo)

La prueba F se utiliza para probar la hipótesis de que **todos los coeficientes del modelo** son cero (es decir, que las variables predictoras no tienen ningún poder explicativo). La hipótesis nula es:

$$
H_0: \beta_1 = \beta_2 = \dots = \beta_p = 0
$$

Contra la hipótesis alternativa:

$$
H_a: \text{Al menos un } \beta_j \neq 0
$$

La estadística de prueba F es:

$$
F = \frac{\text{Variabilidad explicada}}{\text{Variabilidad no explicada}} = \frac{\frac{RSS_0 - RSS_1}{p}}{\frac{RSS_1}{n-p-1}}
$$

Donde:
- $ RSS_0 $ es la suma de los cuadrados de los residuos del modelo sin variables predictoras,
- $ RSS_1 $ es la suma de los cuadrados de los residuos del modelo ajustado,
- $ n $ es el número de observaciones,
- $ p $ es el número de variables predictoras.

Si el valor p de la prueba F es pequeño, se puede concluir que al menos una de las variables predictoras es significativa.

### c. Intervalos de Confianza

Para cada coeficiente $ \beta_j $, se puede construir un intervalo de confianza para estimar el rango dentro del cual se espera que se encuentre el verdadero valor de $ \beta_j $ con un cierto nivel de confianza (por ejemplo, 95%).

El intervalo de confianza se calcula como:

$$
\hat{\beta_j} \pm t_{\alpha/2} \cdot SE(\hat{\beta_j})
$$

Donde $ t_{\alpha/2} $ es el valor crítico de la distribución t para el nivel de confianza deseado.

# 📌 Explicación de Forward, Backward y Stepwise
%%markdown
## 🔹 Selección de Variables: Forward, Backward y Stepwise
La selección de variables es una técnica utilizada en **Regresión Lineal** para encontrar el mejor subconjunto de predictores.

### **1️⃣ Forward Selection (Selección hacia adelante)**
- Se inicia con **ninguna variable en el modelo**.
- Se agregan variables una por una **según su impacto en la métrica objetivo** (e.g., R², AIC, BIC).
- Se detiene cuando **ninguna nueva variable mejora significativamente el modelo**.

### **2️⃣ Backward Elimination (Eliminación hacia atrás)**
- Se inicia con **todas las variables en el modelo**.
- Se eliminan variables una por una **según su insignificancia estadística** (e.g., p-valor > 0.05).
- Se detiene cuando **todas las variables restantes son significativas**.

### **3️⃣ Stepwise Selection (Selección paso a paso)**
- Es una combinación de **Forward y Backward**.
- Se agregan variables como en Forward, pero en cada paso se verifica si alguna variable agregada previamente **deja de ser significativa**.
- Si una variable se vuelve irrelevante, **se elimina** antes de continuar con el siguiente paso.

## 🔹 Muestreo Básico en Machine Learning
El muestreo es una técnica utilizada para **seleccionar una parte de los datos** con el fin de entrenar y evaluar modelos.

### **Tipos Comunes de Muestreo:**
1️⃣ **Muestreo Aleatorio Simple:** Se selecciona una muestra sin sesgo de la población total.  
2️⃣ **Muestreo Estratificado:** Se asegura que la muestra **mantenga la proporción de las clases** (usado en clasificación).  
3️⃣ **Muestreo por Bootstrap:** Se extraen muestras con reemplazo para entrenar modelos robustos.

# 📌  Explicación de Métricas Comunes en Regresión
%%markdown
## 🔹 Métricas Más Comunes en Regresión
Las métricas más utilizadas para evaluar modelos de regresión incluyen:

### **1️⃣ Error Cuadrático Medio (MSE)**
$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$  
Cuantifica el error promedio al **elevar las diferencias al cuadrado** (penaliza más los errores grandes).

### **2️⃣ Error Absoluto Medio (MAE)**
$$ MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i| $$  
Calcula el error medio en valores absolutos.

### **3️⃣ Coeficiente de Determinación (R²)**
$$ R^2 = 1 - \frac{SS_{res}}{SS_{tot}} $$  
Mide qué proporción de la varianza de **y** es explicada por el modelo (valor entre 0 y 1).

---

# 📌  Explicación de la Prueba de Normalidad de Shapiro-Wilk
%%markdown
## 🔹 Prueba de Normalidad de Errores de Shapiro-Wilk
La **prueba de Shapiro-Wilk** evalúa si los residuos de un modelo siguen una distribución normal.

### **Hipótesis:**
- **H₀ (Hipótesis Nula):** Los residuos siguen una distribución normal.
- **H₁ (Hipótesis Alternativa):** Los residuos no son normales.

Si el **p-valor > 0.05**, **NO** se rechaza H₀, lo que significa que los residuos pueden considerarse normales.

---

```python
import itertools
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from statsmodels.api import OLS, add_constant
from scipy.stats import shapiro

# Simulación de datos de una serie de tiempo con fechas
np.random.seed(42)
n_samples = 1500
df = pd.DataFrame({
    'date': pd.date_range(start='2020-01-01', periods=n_samples, freq='D'),
    'X1': np.random.rand(n_samples) * 100,
    'X2': np.random.rand(n_samples) * 50,
    'X3': np.random.rand(n_samples) * 10
})
df['y'] = 3 * df['X1'] + 2 * df['X2'] - 1.5 * df['X3'] + np.random.randn(n_samples) * 5

# Parámetros configurables
time_series_mode = True

train_intervals = [('2020-01-01', '2021-01-01'), ('2020-06-01', '2021-06-01')]
test_intervals = [('2021-01-02', '2022-01-01'), ('2021-06-02', '2022-06-01')]
selection_methods = ['none', 'forward', 'backward', 'stepwise']
sampling_rates = [0.5, 0.7, 1.0]
evaluation_metrics = ['r2', 'mape', 'mse']

# Función para calcular métricas
def compute_metrics(y_true, y_pred):
    if len(y_true) == 0:
        return 'NA', 'NA', 'NA', ['NA'] * 5
    mse = mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    quantiles = np.percentile(np.abs(y_true - y_pred), [5, 30, 50, 70, 95]).tolist()
    return mse, r2, mape, quantiles

# Prueba de normalidad en residuos
def test_normality(residuals):
    stat, p_value = shapiro(residuals)
    return "Sí" if p_value > 0.05 else "No"

# Selección de variables
def select_features(X_train, y_train, method):
    if method == 'none':
        return X_train.columns.tolist()

    features = list(X_train.columns)
    selected_features = []

    if method in ['forward', 'stepwise']:
        for feature in features:
            selected_features.append(feature)
            model = OLS(y_train, add_constant(X_train[selected_features])).fit()
            if method == 'stepwise' and model.pvalues.max() > 0.05:
                selected_features.remove(feature)

    if method in ['backward', 'stepwise']:
        selected_features = features[:]
        while len(selected_features) > 1:
            model = OLS(y_train, add_constant(X_train[selected_features])).fit()
            pvalues = model.pvalues.drop('const', errors='ignore')
            worst_pval = pvalues.idxmax()
            if pvalues[worst_pval] > 0.05 and worst_pval in selected_features:
                selected_features.remove(worst_pval)
            else:
                break

    return selected_features

# Almacenar resultados
results = []

# Generar todas las combinaciones de parámetros
for (train_start, train_end), (test_start, test_end), method, sample_rate in itertools.product(train_intervals, test_intervals, selection_methods, sampling_rates):

    train_data = df[(df['date'] >= train_start) & (df['date'] <= train_end)]
    test_data = df[(df['date'] >= test_start) & (df['date'] <= test_end)]

    train_data = train_data.sample(frac=sample_rate, random_state=42)
    test_data = test_data.sample(frac=sample_rate, random_state=42)

    validation_data = df[~df.index.isin(train_data.index) & ~df.index.isin(test_data.index)]
    if sample_rate == 1.0:
        validation_data = None

    X_train, y_train = train_data.drop(columns=['y', 'date']), train_data['y']
    X_test, y_test = test_data.drop(columns=['y', 'date']), test_data['y']

    if time_series_mode:
        train_total = df[(df['date'] >= train_start) & (df['date'] <= train_end)]
        test_total = df[(df['date'] >= test_start) & (df['date'] <= test_end)]
    else:
        train_total, test_total = None, None

    selected_features = select_features(X_train, y_train, method)
    X_train, X_test = X_train[selected_features], X_test[selected_features]

    model = LinearRegression()
    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    y_validation_pred = model.predict(validation_data[selected_features]) if validation_data is not None else None
    y_train_total_pred = model.predict(train_total[selected_features]) if train_total is not None else None
    y_test_total_pred = model.predict(test_total[selected_features]) if test_total is not None else None
    y_total_pred = model.predict(df[selected_features])

    mse_train, r2_train, mape_train, quantiles_train = compute_metrics(y_train, y_train_pred)
    mse_test, r2_test, mape_test, quantiles_test = compute_metrics(y_test, y_test_pred)
    mse_validation, r2_validation, mape_validation, quantiles_validation = compute_metrics(validation_data['y'], y_validation_pred) if validation_data is not None else ('NA', 'NA', 'NA', ['NA'] * 5)
    mse_train_total, r2_train_total, mape_train_total, quantiles_train_total = compute_metrics(train_total['y'], y_train_total_pred) if train_total is not None else ('NA', 'NA', 'NA', ['NA'] * 5)
    mse_test_total, r2_test_total, mape_test_total, quantiles_test_total = compute_metrics(test_total['y'], y_test_total_pred) if test_total is not None else ('NA', 'NA', 'NA', ['NA'] * 5)
    mse_total, r2_total, mape_total, quantiles_total = compute_metrics(df['y'], y_total_pred)

    normality_train = test_normality(y_train - y_train_pred)

    results.append({
        'id': len(results) + 1,
        'train_start': train_start, 'train_end': train_end,
        'test_start': test_start, 'test_end': test_end,
        'method': method, 'sample_rate': sample_rate, 'metric_optimized': metric,
        'num_train': len(y_train), 'num_test': len(y_test), 'num_validation': len(validation_data) if validation_data is not None else 'NA',
        'normality_train': normality_train,
        'mse_train': mse_train, 'r2_train': r2_train, 'mape_train': mape_train, 'quantiles_train': quantiles_train,
        'mse_test': mse_test, 'r2_test': r2_test, 'mape_test': mape_test, 'quantiles_test': quantiles_test,
        'mse_validation': mse_validation, 'r2_validation': r2_validation, 'mape_validation': mape_validation, 'quantiles_validation': quantiles_validation,
        'mse_total': mse_total, 'r2_total': r2_total, 'mape_total': mape_total, 'quantiles_total': quantiles_total
    })

df_results = pd.DataFrame(results)

pd.set_option('display.max_columns', None)  # Mostrar todas las columnas
pd.set_option('display.width', 1000)  # Ajustar el ancho de la salida


# Guardar en un archivo CSV para análisis posterior
df_results.to_csv("resultados_regresion.csv", index=False)

print("Los resultados se han guardado en 'resultados_regresion.csv'. Puedes descargarlos y analizarlos.")


df_results.head(5)
```

## 🔹 Regularización en Regresión Lineal: Ridge y Lasso

La **Regresión Lineal Múltiple** se expresa como:

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_p x_p + \epsilon
$$

Donde:
- $ y $ es la variable dependiente.
- $ x_i $ son las variables independientes.
- $ \beta_i $ son los coeficientes del modelo.
- $ \epsilon $ es el término de error.

Sin embargo, cuando existe **multicolinealidad** o demasiadas variables, el modelo puede sobreajustarse. Para evitarlo, se usan técnicas de **regularización** como **Ridge y Lasso**.

---

## 🔹 1️⃣ Regresión Ridge (Regularización L2)
La regresión **Ridge** agrega una penalización sobre la norma **L2** de los coeficientes:

$$
\min_{\beta} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} \beta_j^2
$$

Donde:
- $ \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $ es el **Error Cuadrático Medio (MSE)**.
- $ \alpha \sum_{j=1}^{p} \beta_j^2 $ es el **término de penalización**.
- $ \alpha $ controla la penalización:
  - **Si $\alpha = 0$**, se obtiene la regresión lineal estándar.
  - **Si $\alpha$ es grande**, se reducen los coeficientes $\beta_j$, pero nunca se eliminan.

✅ **Ventaja:** Reduce la varianza y mejora la estabilidad del modelo.  
❌ **Desventaja:** No permite seleccionar variables (todos los coeficientes siguen siendo distintos de cero).

---

## 🔹 2️⃣ Regresión Lasso (Regularización L1)
La regresión **Lasso** introduce la penalización **L1**:

$$
\min_{\beta} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{p} |\beta_j|
$$

- En lugar de elevar al cuadrado los coeficientes, Lasso usa **valores absolutos**.
- **Diferencia clave:** Algunos coeficientes $\beta_j$ pueden volverse **exactamente cero**, lo que significa que Lasso **hace selección de variables**.

✅ **Ventaja:** Realiza selección de variables automáticamente, eliminando predictores irrelevantes.  
❌ **Desventaja:** Puede generar inestabilidad cuando hay muchas variables correlacionadas.

---

## 🔹 3️⃣ Comparación entre Ridge y Lasso
| Característica  | Ridge (L2)  | Lasso (L1)  |
|---------------|------------|------------|
| **Objetivo**  | Reducir la varianza del modelo | Seleccionar variables relevantes |
| **Penalización**  | $ \alpha \sum \beta_j^2 $ | $ \alpha \sum |\beta_j| $ |
| **Coeficientes**  | Se reducen pero **nunca son cero** | Algunos se vuelven exactamente **cero** |
| **Ideal para**  | Datos con multicolinealidad | Modelos donde se quiere selección de variables |

---


```python
import itertools
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge, Lasso
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score
from statsmodels.api import OLS, add_constant
from scipy.stats import shapiro

# Simulación de datos de una serie de tiempo con fechas
np.random.seed(42)
n_samples = 1500
df = pd.DataFrame({
    'date': pd.date_range(start='2020-01-01', periods=n_samples, freq='MS'),  # Intervalo mensual en lugar de días
    'X1': np.random.rand(n_samples) * 100,
    'X2': np.random.rand(n_samples) * 50,
    'X3': np.random.rand(n_samples) * 10
})
df['y'] = 3 * df['X1'] + 2 * df['X2'] - 1.5 * df['X3'] + np.random.randn(n_samples) * 5

# Parámetros configurables
time_series_mode = True

train_intervals = [('2020-01', '2021-06'), ('2021-01', '2022-06')]  # Ahora se usa año-mes
test_intervals = [('2022-07', '2023-06'), ('2023-01', '2023-12')]
selection_methods = ['none', 'forward', 'backward', 'stepwise']
sampling_rates = [0.5, 0.7, 1.0]

# Configuración de GridSearch para Ridge y Lasso
param_grid = {'alpha': np.logspace(-3, 3, 10)}

# Función para calcular métricas
def compute_metrics(y_true, y_pred):
    if len(y_true) == 0:
        return 'NA', 'NA', 'NA', ['NA'] * 5
    mse = mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    quantiles = np.percentile(np.abs(y_true - y_pred), [5, 30, 50, 70, 95]).tolist()
    return mse, r2, mape, quantiles

# Prueba de normalidad en residuos
def test_normality(residuals):
    stat, p_value = shapiro(residuals)
    return "Sí" if p_value > 0.05 else "No"

# Selección de variables
def select_features(X_train, y_train, method):
    if method == 'none':
        return X_train.columns.tolist()

    features = list(X_train.columns)
    selected_features = []

    if method in ['forward', 'stepwise']:
        for feature in features:
            selected_features.append(feature)
            model = OLS(y_train, add_constant(X_train[selected_features])).fit()
            if method == 'stepwise' and model.pvalues.max() > 0.05:
                selected_features.remove(feature)

    if method in ['backward', 'stepwise']:
        selected_features = features[:]
        while len(selected_features) > 1:
            model = OLS(y_train, add_constant(X_train[selected_features])).fit()
            pvalues = model.pvalues.drop('const', errors='ignore')
            worst_pval = pvalues.idxmax()
            if pvalues[worst_pval] > 0.05 and worst_pval in selected_features:
                selected_features.remove(worst_pval)
            else:
                break

    return selected_features

# Almacenar resultados
results = []

# Generar todas las combinaciones de parámetros
for (train_start, train_end), (test_start, test_end), method, sample_rate in itertools.product(train_intervals, test_intervals, selection_methods, sampling_rates):

    train_data = df[(df['date'] >= train_start) & (df['date'] <= train_end)]
    test_data = df[(df['date'] >= test_start) & (df['date'] <= test_end)]

    train_data = train_data.sample(frac=sample_rate, random_state=42)
    test_data = test_data.sample(frac=sample_rate, random_state=42)

    validation_data = df[~df.index.isin(train_data.index) & ~df.index.isin(test_data.index)]
    if sample_rate == 1.0:
        validation_data = None

    X_train, y_train = train_data.drop(columns=['y', 'date']), train_data['y']
    X_test, y_test = test_data.drop(columns=['y', 'date']), test_data['y']

    if time_series_mode:
        train_total = df[(df['date'] >= train_start) & (df['date'] <= train_end)]
        test_total = df[(df['date'] >= test_start) & (df['date'] <= test_end)]
    else:
        train_total, test_total = None, None

    selected_features = select_features(X_train, y_train, method)
    X_train, X_test = X_train[selected_features], X_test[selected_features]

    # GridSearch para Ridge
    ridge = GridSearchCV(Ridge(), param_grid, cv=5, scoring='r2')
    ridge.fit(X_train, y_train)
    best_ridge = Ridge(alpha=ridge.best_params_['alpha']).fit(X_train, y_train)

    # GridSearch para Lasso
    lasso = GridSearchCV(Lasso(), param_grid, cv=5, scoring='r2')
    lasso.fit(X_train, y_train)
    best_lasso = Lasso(alpha=lasso.best_params_['alpha']).fit(X_train, y_train)

    # Usamos el mejor modelo (Ridge o Lasso según GridSearch)
    model = best_ridge if ridge.best_score_ > lasso.best_score_ else best_lasso

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    y_validation_pred = model.predict(validation_data[selected_features]) if validation_data is not None else None
    y_total_pred = model.predict(df[selected_features])

    mse_train, r2_train, mape_train, quantiles_train = compute_metrics(y_train, y_train_pred)
    mse_test, r2_test, mape_test, quantiles_test = compute_metrics(y_test, y_test_pred)
    mse_validation, r2_validation, mape_validation, quantiles_validation = compute_metrics(validation_data['y'], y_validation_pred) if validation_data is not None else ('NA', 'NA', 'NA', ['NA'] * 5)
    mse_total, r2_total, mape_total, quantiles_total = compute_metrics(df['y'], y_total_pred)

    normality_train = test_normality(y_train - y_train_pred)

    results.append({
        'id': len(results) + 1,
        'train_start': train_start, 'train_end': train_end,
        'test_start': test_start, 'test_end': test_end,
        'method': method, 'sample_rate': sample_rate,
        'best_model': 'Ridge' if ridge.best_score_ > lasso.best_score_ else 'Lasso',
        'best_alpha': ridge.best_params_['alpha'] if ridge.best_score_ > lasso.best_score_ else lasso.best_params_['alpha'],
        'num_train': len(y_train), 'num_test': len(y_test), 'num_validation': len(validation_data) if validation_data is not None else 'NA',
        'normality_train': normality_train,
        'mse_train': mse_train, 'r2_train': r2_train, 'mape_train': mape_train,
        'mse_test': mse_test, 'r2_test': r2_test, 'mape_test': mape_test,
        'mse_validation': mse_validation, 'r2_validation': r2_validation, 'mape_validation': mape_validation,
        'mse_total': mse_total, 'r2_total': r2_total, 'mape_total': mape_total
    })

df_results = pd.DataFrame(results)
df_results.to_csv("resultados_regresion.csv", index=False)

print("Resultados guardados en 'resultados_regresion.csv'")
df_results.head(5)

```
